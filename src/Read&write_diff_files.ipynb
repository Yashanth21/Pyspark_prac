```python
employee_data1 = [
    (101,"Rahul","IT",60000),
    (102,"Priya","HR",50000),
    (103,"Arjun","Finance",70000),
    (104,"Sneha","IT",65000),
    (105,"Ravi","Sales",45000)
]

columns = ["EmployeeID","EmployeeName","Department","Salary"]

employee1_df = spark.createDataFrame(employee_data1, columns)
```

```python
from pyspark.sql.functions import col, when
```

```python
bonus_df = employee1_df.withColumn("Bonus", col("Salary")*0.10)

bonus_df.show()
```

```python
grade_df = bonus_df.withColumn(
    "Grade",
    when(col("Salary")>=65000,"A")
    .when(col("Salary")>=55000,"B")
    .otherwise("C")
)

grade_df.show()
```

```python
rename_df = grade_df.withColumnRenamed("Salary","EmployeeSalary")

rename_df.show()
```

```python
rename_df = grade_df.withColumnRenamed("Salary","EmployeeSalary")

rename_df.show()
```

```python
employee_data2 = [
    (101,"Rahul","IT",60000),
    (102,"Priya","HR",50000),
    (103,"Arjun","Finance",70000),
    (104,"Sneha","IT",65000),
    (105,"Ravi","Sales",45000),
    (106,"Anjali","HR",52000),
    (107,"Vikram","IT",72000),
    (108,"Neha","Finance",68000)
]

columns = ["EmployeeID","EmployeeName","Department","Salary"]

employee_df2 = spark.createDataFrame(employee_data2, columns)
```

```python
from pyspark.sql.functions import sum, avg, max, min, count
```

```python
employee_df2.groupBy("Department").count().show()
```

```python
employee_df2.groupBy("Department").avg("Salary").show()
```

```python
employee_df2.groupBy("Department").agg(
    count("*").alias("Employees"),
    avg("Salary").alias("AverageSalary"),
    max("Salary").alias("HighestSalary"),
    min("Salary").alias("LowestSalary"),
    sum("Salary").alias("TotalSalary")
).show()
```
