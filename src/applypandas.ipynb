```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
import pandas as pd
```

```python
employee_data = [

    (101,"Rahul","IT",60000),
    (102,"Priya","HR",45000),
    (103,"Arjun","IT",70000),
    (104,"Sneha","Finance",85000),
    (105,"Ravi","HR",52000),
    (106,"Anita","IT",95000),
    (107,"John","Sales",40000),
    (108,"Sara","Marketing",78000),
    (109,"Rohit","Finance",65000),
    (110,"Megha","IT",88000),
    (111,"Kiran","IT",92000),
    (112,"David","Sales",55000),
    (113,"Neha","Finance",73000),
    (114,"Amit","Marketing",68000),
    (115,"Pooja","HR",58000)

]

employee_df = spark.createDataFrame(
    employee_data,
    ["ID","Name","Department","Salary"]
)

employee_df.show()
```

```python
schema = StructType([

    StructField("ID", IntegerType()),
    StructField("Name", StringType()),
    StructField("Department", StringType()),
    StructField("Salary", IntegerType()),

    StructField("DepartmentAverage", DoubleType()),
    StructField("DepartmentMaximum", IntegerType()),
    StructField("DepartmentMinimum", IntegerType()),
    StructField("EmployeeCount", IntegerType()),

    StructField("Bonus", DoubleType()),
    StructField("Rating", StringType())

])
```

```python
def department_statistics(pdf):

    average_salary = pdf["Salary"].mean()

    maximum_salary = pdf["Salary"].max()

    minimum_salary = pdf["Salary"].min()

    employee_count = len(pdf)

    pdf["DepartmentAverage"] = average_salary

    pdf["DepartmentMaximum"] = maximum_salary

    pdf["DepartmentMinimum"] = minimum_salary

    pdf["EmployeeCount"] = employee_count

    pdf["Bonus"] = pdf["Salary"] * 0.10

    pdf["Rating"] = pdf["Salary"].apply(

        lambda x:
            "Excellent"
            if x >= average_salary
            else "Average"

    )

    return pdf
```

```python
result_df = employee_df.groupBy("Department").applyInPandas(
    department_statistics,
    schema
)
```

```python
result_df.orderBy("Department","Salary").show()
```
