```python
from pyspark.sql.functions import *

employee_data = [
    (101, "Rahul", 60000.75, 82, 76, 91),
    (102, "Priya", -50000.40, 90, 95, 88),
    (103, "Arjun", 70000.90, 75, 80, 79),
    (104, "Sneha", 65000.35, 88, 84, 86),
    (105, "Ravi", -45000.99, 70, 65, 72)
]

columns = [
    "EmployeeID",
    "EmployeeName",
    "Salary",
    "Maths",
    "Science",
    "English"
]

employee_df = spark.createDataFrame(employee_data, columns)

employee_df.show()

result_df = employee_df.select(
    "EmployeeID",
    "EmployeeName",
    "Salary",

    abs("Salary").alias("AbsoluteSalary"),

    round("Salary", 1).alias("RoundedSalary"),

    ceil("Salary").alias("CeilingSalary"),

    floor("Salary").alias("FloorSalary"),

    greatest("Maths", "Science", "English").alias("HighestMarks")
)

result_df.show()
```

```python
from pyspark.sql.functions import *

employee_data = [
    (101,"Rahul","IT",60000),
    (102,"Priya","HR",50000),
    (103,"Arjun","IT",70000),
    (104,"Sneha","Finance",65000),
    (105,"Ravi","HR",55000),
    (106,"Anjali","Finance",75000)
]

columns = ["EmployeeID","EmployeeName","Department","Salary"]

employee_df = spark.createDataFrame(employee_data, columns)

employee_df.show()

result_df = employee_df.groupBy("Department").agg(
    count("*").alias("EmployeeCount"),
    sum("Salary").alias("TotalSalary"),
    avg("Salary").alias("AverageSalary"),
    min("Salary").alias("MinimumSalary"),
    max("Salary").alias("MaximumSalary")
)

result_df.show()
```
