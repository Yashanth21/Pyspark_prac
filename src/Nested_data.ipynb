```python
from pyspark.sql.functions import *

data = [
    ("Rahul", "Bangalore", "Karnataka", 560001),
    ("Priya", "Hyderabad", "Telangana", 500001)
]

columns = ["Customer", "City", "State", "Pincode"]

df = spark.createDataFrame(data, columns)

struct_df = df.select(
    "Customer",
    struct("City", "State", "Pincode").alias("Address")
)

struct_df.show(truncate=False)
```

**Above is example for Struct usage**

```python
from pyspark.sql.functions import *

data = [
    ("Rahul", ["Laptop", "Mouse", "Keyboard"]),
    ("Priya", ["Mobile", "Earphones"])
]

columns = ["Customer", "Products"]

df = spark.createDataFrame(data, columns)

df.show(truncate=False)
```

Below for explode


```python
from pyspark.sql.functions import *

data = [
    ("Rahul", ["Laptop", "Mouse", "Keyboard"]),
    ("Priya", ["Mobile", "Earphones"])
]

columns = ["Customer", "Products"]

df = spark.createDataFrame(data, columns)

df.select(
    "Customer",
    explode("Products").alias("Product")
).show()
```
