```python
from pyspark.sql.functions import *

employee_data = [
    (101," rahul kumar ","it","bangalore"),
    (102," PRIYA ","hr","Hyderabad"),
    (103,"Arjun","Finance","Chennai"),
    (104,"Sneha","IT","PUNE"),
    (105,"Ravi Kumar","Sales","mumbai")
]

columns = ["EmployeeID","EmployeeName","Department","City"]

employee_df = spark.createDataFrame(employee_data, columns)

print("Original Data")
employee_df.show(truncate=False)

clean_df = employee_df \
    .withColumn("EmployeeName", trim(col("EmployeeName"))) \
    .withColumn("EmployeeName", initcap(col("EmployeeName"))) \
    .withColumn("Department", upper(col("Department"))) \
    .withColumn("City", initcap(col("City"))) \
    .withColumn("NameLength", length(col("EmployeeName")))

print("Cleaned Data")
clean_df.show(truncate=False)
```
