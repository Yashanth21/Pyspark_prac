```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
employee_data = [

    (101,"rahul","IT",60000),
    (102,"priya","HR",45000),
    (103,"arjun","IT",70000),
    (104,"sneha","Finance",85000),
    (105,"ravi","HR",52000),
    (106,"anita","IT",95000),
    (107,"john","Sales",40000),
    (108,"sara","Marketing",78000),
    (109,"rohit","Finance",65000),
    (110,"megha","IT",88000)

]

employee_df = spark.createDataFrame(
    employee_data,
    ["ID","Name","Department","Salary"]
)
```

```python
def grade(salary):

    if salary < 50000:
        return "C"

    elif salary < 80000:
        return "B"

    else:
        return "A"
```

```python
grade_udf = udf(
    grade,
    StringType()
)
```

```python
udf_df = employee_df \
.withColumn("EmployeeName", initcap(col("Name"))) \
.withColumn("AnnualSalary", col("Salary") * 12) \
.withColumn("Bonus", col("Salary") * 0.10) \
.withColumn("Grade", grade_udf(col("Salary"))) \
.withColumn("Department", upper(col("Department")))
```
